## <span style='color:blue'>Rubric to calculate monthly loan payments (with compound interest)</span>

### Loan details

#### Enter data here:

```python
loan = 20000            # dollars ($)
rate = 5.9              # annual percentage (interest) rate (%))
years = 5               # length of loan in years
```

### Background math

```python
n = 12                  # number of months 
R = rate * 0.01 / n     # monthly interest rate in decimals (5.9% = 0.059)
m = years * n           # length of loan in months
```

### Calculate monthly payment

```python
payment = (loan * R) / (1 - (1 + R)**(-m))
payment = round(payment,2)
print(payment)
```

## <span style='color:blue'>Rubric to calculate maximum loan given a target monthly payment</span>

### Desired monthly payment and loan details (APR, years)

#### Enter data here:

```python
payment = 385.73        # maximum monthly payment ($)
rate = 5.9              # annual percentage (interest) rate (%)
years = 5               # length of loan in years
```

### Background math

```python
n = 12                  # number of months
R = rate * 0.01 / n     # monthly interest rate in decimals
m = years * n           # length of loan in months
```

### Calculate target loan amount given loan details (APR, years)

```python
loan = payment * (1 - (1 + R)**(-m)) / R
loan = round(loan,0)
print(loan)
```
